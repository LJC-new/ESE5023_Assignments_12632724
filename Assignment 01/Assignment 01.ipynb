```python
a = float(input("请输入 a : "))
b = float(input("请输入 b : "))
c = float(input("请输入 c : "))
def Print_values(a, b, c):
    if a > b:
        if b > c:
            print(f"{a}, {b}, {c}")
        else:  # b <= c
            if a > c:
                print(f"{a}, {c}, {b}")
            else:  # a <= c
                print(f"{c}, {a}, {b}")
    else:  # a <= b
        if b > c:
            print(f"{c}, {a}, {b}")
        else:  # b <= c
            print(f"{c}, {b}, {a}")
Print_values(a, b, c)
```

```python
import random

M1 = []
for i in range(5):
    row = []
    for j in range(10):
        row.append(random.randint(0, 50))
    M1.append(row)
    
M2 = []
for i in range(10):
    row = []
    for j in range(5):
        row.append(random.randint(0, 50))
    M2.append(row)

print("M1 (5x10):")
for row in M1:
    print(row)

print("\nM2 (10x5):")
for row in M2:
    print(row)
    
def Matrix_multip(A, B):
    m = len(A)          
    n = len(B)          
    p = len(B[0])      
    result = []
    for i in range(m):
        row = []
        for j in range(p):
            row.append(0)
        result.append(row)
    for i in range(m):
        for j in range(p):         
            total = 0
            for k in range(n):      
                total = total + A[i][k] * B[k][j]
            result[i][j] = total

    return result

M3 = Matrix_multip(M1, M2)
print("\nM1 * M2 (结果为 5x5):")
for row in M3:
    print(row)
```

```python
def Pascal_triangle(k):
    hang = [1]
    for i in range(1, k + 1):
        xin_hang = [1]
        for j in range(1, i):
            xin_hang.append(hang[j - 1] + hang[j])
        xin_hang.append(1)
        hang = xin_hang
    return hang
print(Pascal_triangle(100))

print(Pascal_triangle(200))
```

```python
x = int(input("请输入一个0-100的整数 ："))
def Least_moves(x):
    n = x          
    moves = 0    
    while n > 1:               
        if n % 2 == 0:         
            n = n // 2
        else:                  
            n = n - 1
        moves = moves + 1      
    print(moves)  
Least_moves(x)
```

```python
def Find_expression(target):
    digits = [1, 2, 3, 4, 5, 6, 7, 8, 9]
    results = []
    for code in range(3 ** 8):
        choice = []
        c = code
        for _ in range(8):
            choice.append(c % 3)
            c = c // 3
        expr = str(digits[0])
        value = digits[0]
        last = digits[0]
        for i in range(8):
            d = digits[i + 1]
            if choice[i] == 0:              
                expr = expr + "+" + str(d)
                value = value + d
                last = d
            elif choice[i] == 1:            
                expr = expr + "-" + str(d)
                value = value - d
                last = -d
            else:                            
                expr = expr + str(d)
                if last >= 0:
                    new_last = last * 10 + d
                else:
                    new_last = last * 10 - d
                value = value - last + new_last
                last = new_last
        if value == target:
            results.append(expr)
    for expr in results:
        print(expr + "=" + str(target))
    return results
Find_expression(50)
```

```python
import matplotlib.pyplot as plt
def Find_expression(target, show=False):
    digits = [1, 2, 3, 4, 5, 6, 7, 8, 9]
    results = []

    for code in range(3 ** 8):
        choice = []
        c = code
        for _ in range(8):
            choice.append(c % 3)
            c = c // 3

        expr = str(digits[0])
        value = digits[0]
        last = digits[0]

        for i in range(8):
            d = digits[i + 1]
            if choice[i] == 0:              
                expr = expr + "+" + str(d)
                value = value + d
                last = d
            elif choice[i] == 1:            
                expr = expr + "-" + str(d)
                value = value - d
                last = -d
            else:                            
                expr = expr + str(d)
                if last >= 0:
                    new_last = last * 10 + d
                else:
                    new_last = last * 10 - d
                value = value - last + new_last
                last = new_last

        if value == target:
            results.append(expr)

    if show:
        for expr in results:
            print(expr + "=" + str(target))

    return results

Total_solutions = []
for i in range(1, 101):
    Total_solutions.append(len(Find_expression(i)))

print("Total_solutions =")
print(Total_solutions)
print()

max_count = max(Total_solutions)
min_count = min(Total_solutions)

max_numbers = []
min_numbers = []
for idx in range(100):
    if Total_solutions[idx] == max_count:
        max_numbers.append(idx + 1)
    if Total_solutions[idx] == min_count:
        min_numbers.append(idx + 1)

print("最大解数量：", max_count)
print("对应的数：", max_numbers)
print()
print("最小解数量：", min_count)
print("对应的数：", min_numbers)

plt.figure(figsize=(15, 6))
plt.bar(range(1, 101), Total_solutions, color="steelblue", edgecolor="black")
plt.xlabel("Integer i (1 ~ 100)")
plt.ylabel("Number of solutions")
plt.title("Total_solutions: number of solutions for each integer 1~100")
plt.xticks(range(1, 101, 5))
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```
